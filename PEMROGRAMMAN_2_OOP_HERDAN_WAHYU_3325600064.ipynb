```python
import os

class DataRecord:
    def __init__(self, content):
        self.content = content

    def __str__(self):
        return str(self.content)

class Dataset:
    def __init__(self, name):
        """
        Untuk inisialisasi nilai awal pada object Dataset.
        Dengan Atribut
        records: Kumpulan data penjualan (list of DataRecord)
        name: Identifier dataset
        """
        self.records = []
        self.name = name

    def load_data_from_file(self, filename):
        """
        Membaca isi file .txt dan menyimpan isi tiap barisnya ke dalam records.
        """
        if not os.path.exists(filename):
            print(f"File {filename} tidak ditemukan.")
            return None

        try:
            with open(filename, 'r') as file:
                lines = file.readlines()
                count = 0
                for line in lines:
                    clean_data = line.strip()
                    if clean_data:
                        new_record = DataRecord(clean_data)
                        self.records.append(new_record)
                        count += 1
                return count
        except Exception as e:
            print(f"Terjadi kesalahan: {e}")
            return None

    def add_data(self, record):
        """
        Menambahkan data penjualan baru ke dalam records.
        Return True bila berhasil, False bila bukan object DataRecord.
        """
        if isinstance(record, DataRecord):
            self.records.append(record)
            return True
        else:
            return False

    def display_data(self):
        """
        Menampilkan semua data penjualan yang tersimpan pada records.
        """
        if not self.records:
            print("Tidak ada data bila records kosong.")
        else:
            print(f"--- Data pada Dataset: {self.name} ---")
            for i, rec in enumerate(self.records, 1):
                print(f"{i}. {rec}")


# Membuat file dummy
with open("data_penjualan.txt", "w") as f:
    f.write("Kopi Susu - 15000\nRoti Bakar - 20000\nTeh Tarik - 12000")

# Inisialisasi Dataset
my_dataset = Dataset("Penjualan Kafe")

# Load data
jumlah = my_dataset.load_data_from_file("data_penjualan.txt")
print(f"Berhasil memuat {jumlah} data.\n")

# Add data manual
new_item = DataRecord("Pisang Goreng - 10000")
my_dataset.add_data(new_item)

# Show output
my_dataset.display_data()
```

```python
import os

class DataRecord:
    def __init__(self, name, qty=None, price=None):
        self.name = name
        self.qty = qty
        self.price = price

    def __str__(self):
        if self.qty is not None and self.price is not None:
            return f"Name: {self.name}; Qty: {self.qty}; Price: {self.price:,}"
        return self.name

class Dataset:
    def __init__(self, name):
        self.records = []
        self.name = name

    def load_data_from_file(self, filename):
        if not os.path.exists(filename):
            print(f"File {filename} is not found")
            return None

        try:
            with open(filename, 'r') as file:
                lines = file.readlines()
                count = 0
                for line in lines:
                    clean_data = line.strip()
                    if clean_data:
                        # Simulasi parsing data sederhana dari file
                        parts = clean_data.split(',')
                        if len(parts) == 3:
                            new_rec = DataRecord(parts[0], int(parts[1]), int(parts[2]))
                        else:
                            new_rec = DataRecord(clean_data)
                        self.records.append(new_rec)
                        count += 1
                print(f"{count} from {filename} added to {self.name}")
                return count
        except Exception as e:
            return None

    def add_data(self, record):
        if isinstance(record, DataRecord):
            self.records.append(record)
            print(f"{record.name} added to {self.name}")
            return True
        else:
            print(f"{record} is not object of DataRecord")
            return False

    def display_data(self):
        print(f"\n{self.name}:")
        if not self.records:
            print("Tidak ada data.")
        else:
            for i, rec in enumerate(self.records, 1):
                print(f"{i}. {rec}")

# Membuat file data_penjualan.txt agar bisa saya load
with open("data_penjualan.txt", "w") as f:
    f.write("Laptop,2,7000000\nMouse,5,200000\nKeyboard,3,350000\nMonitor,1,2500000")

dataset_alpha = Dataset("dataset_alpha")
dataset_beta = Dataset("dataset_beta")

# Add data_record_1, 2, dan 3 ke dataset_alpha
dr1 = DataRecord("Buku tulis", 11, 25500)
dr2 = DataRecord("Sticky notes", 5, 8000)
dr3 = DataRecord("Rautan pensil", 3, 4000)

dataset_alpha.add_data(dr1)
dataset_alpha.add_data(dr1)
dataset_alpha.add_data(dr2)
dataset_alpha.add_data(dr3)

# Add string "Correction tape" (Ini harusnya gagal karena bukan object DataRecord)
dataset_alpha.add_data("Correction Tape")

# isi file data_penjualan.txt ke dataset_beta
dataset_beta.load_data_from_file("data_penjualan.txt")

# Tambahkan file yang tidak tersedia ke dataset_beta
dataset_beta.load_data_from_file("penjualan.txt")

# Show dua dataset
dataset_alpha.display_data()
dataset_beta.display_data()
```

```python
class DataAnalyzer:
    def __init__(self, dataset):
        """
        Inisialisasi nilai awal pada object DataAnalyzer.
        Atributnya hanya dataset yang akan dianalisis.
        """
        self.dataset = dataset

    def calculate_total_sales(self, include_tax=False):
        """
        Menghitung total penjualan semua produk pada dataset.
        menggunakan parameter include_tax untuk menentukan apakah pajak 10% harus dihitung.
        dengan return Total penjualan (float).
        """
        if not self.dataset.records:
            return None

        total = 0
        for record in self.dataset.records:
            item_total = record.qty * record.price
            total += item_total

        if include_tax:
            total = total * 1.1

        return total

    def find_best_selling(self):
        """
        Mencari produk dengan kuantitas penjualan terbesar pada dataset.
        Return: Nama produk dengan kuantitas terbesar (str).
        """
        if not self.dataset.records:
            return None

        best_record = self.dataset.records[0]
        for record in self.dataset.records:
            if record.qty > best_record.qty:
                best_record = record

        return best_record.name

    def calculate_average_price(self):
        """
        Menghitung rata-rata harga produk pada dataset.
        Return: Harga rata-rata produk (float).
        """
        if not self.dataset.records:
            return None

        total_price = 0
        for record in self.dataset.records:
            total_price += record.price

        return total_price / len(self.dataset.records)

analyzer = DataAnalyzer(dataset_beta)
total_tanpa_pajak = analyzer.calculate_total_sales(include_tax=False)
total_dengan_pajak = analyzer.calculate_total_sales(include_tax=True)
produk_terlaris = analyzer.find_best_selling()
rata_harga = analyzer.calculate_average_price()

print(f"Analisis untuk {dataset_beta.name}:")
print(f"- Total Penjualan (Tanpa Pajak): {total_tanpa_pajak:,.0f}")
print(f"- Total Penjualan (Termasuk Pajak 10%): {total_dengan_pajak:,.0f}")
print(f"- Produk Terlaris: {produk_terlaris}")
print(f"- Rata-rata Harga Produk: {rata_harga:,.2f}")
```

```python
analyzer_alpha = DataAnalyzer(dataset_alpha)
analyzer_beta = DataAnalyzer(dataset_beta)
total_alpha = analyzer_alpha.calculate_total_sales(include_tax=False)

best_alpha = analyzer_alpha.find_best_selling()

avg_alpha = analyzer_alpha.calculate_average_price()

total_beta = analyzer_beta.calculate_total_sales(include_tax=True)

best_beta = analyzer_beta.find_best_selling()

avg_beta = analyzer_beta.calculate_average_price()

print(f"Analysis of {dataset_alpha.name}")
print(f"Total sales (Without tax): {total_alpha:,.0f}")
print(f"Best selling item: {best_alpha}")
print(f"Average price: {avg_alpha:,.0f}")

print(f"Analysis of {dataset_beta.name}")
print(f"Total sales (With tax): {total_beta:,.0f}")
print(f"Best selling item: {best_beta}")
print(f"Average price: {avg_beta:,.0f}")
```
